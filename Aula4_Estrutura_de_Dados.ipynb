Desafio 1:O simulador "Desfazer"

```python
pilha = []

while True:
    print("\n===== SIMULADOR DESFAZER =====")
    print("1 - Inserir ação")
    print("2 - Desfazer última ação")
    print("3 - Mostrar ações")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        acao = input("Digite a ação realizada: ")
        pilha.append(acao)

        print("Ação adicionada com sucesso!")

    elif opcao == "2":
        if len(pilha) > 0:
            ultima_acao = pilha.pop()
            print("Desfazendo:", ultima_acao)
        else:
            print("Não existem ações para desfazer!")

    elif opcao == "3":
        if len(pilha) == 0:
            print("Nenhuma ação realizada.")
        else:
            print("\nAções realizadas:")

            # Mostra da última ação para a primeira
            for acao in reversed(pilha):
                print("-", acao)

    elif opcao == "0":
        print("Programa encerrado!")
        break

    else:
        print("Opção inválida!")
```

Desafio 2 Sistema de Impressão:

```python
fila = []

while True:
    print("\n===== SISTEMA DE IMPRESSÃO =====")
    print("1 - Adicionar documento")
    print("2 - Imprimir documento")
    print("3 - Mostrar fila")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        documento = input("Digite o nome do documento: ")
        fila.append(documento)

        print("Documento adicionado à fila!")

    elif opcao == "2":
        if len(fila) > 0:
            documento = fila.pop(0)
            print("Imprimindo:", documento)
        else:
            print("A fila está vazia!")

    elif opcao == "3":
        if len(fila) == 0:
            print("A fila está vazia!")
        else:
            print("\nDocumentos aguardando impressão:")

            for documento in fila:
                print("-", documento)

    elif opcao == "0":
        print("Programa encerrado!")
        break

    else:
        print("Opção inválida!")
```

Desafio 3 e Master/Triagem Hospitalar:

```python
from collections import deque

fila = deque()

while True:
    print("\n===== TRIAGEM HOSPITALAR =====")
    print("1 - Adicionar paciente normal")
    print("2 - Adicionar paciente prioritário")
    print("3 - Atender paciente")
    print("4 - Mostrar fila")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        paciente = input("Nome do paciente: ")

        # Adiciona o nome ao final da fila
        fila.append(paciente)

        print("Paciente normal adicionado ao final da fila!")

    elif opcao == "2":
        paciente = input("Nome do paciente prioritário: ")

        # Adiciona o nome ao início da fila
        fila.appendleft(paciente)

        print("Paciente prioritário adicionado ao início da fila!")

    elif opcao == "3":
        if len(fila) > 0:
            paciente = fila.popleft()
            print("Atendendo:", paciente)
        else:
            print("Não existem pacientes na fila!")

    elif opcao == "4":
        if len(fila) == 0:
            print("A fila está vazia!")
        else:
            print("\n===== FILA DE ATENDIMENTO =====")

            for i, paciente in enumerate(fila, start=1):
                print(f"{i}º - {paciente}")

    elif opcao == "0":
        print("Programa encerrado!")
        break

    else:
        print("Opção inválida!")
```
